```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.ml.feature import VectorAssembler, StringIndexer
from pyspark.ml.regression import LinearRegression
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.classification import DecisionTreeClassifier
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.clustering import KMeans
from pyspark.ml.evaluation import RegressionEvaluator
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
```

```python
spark = SparkSession.builder \
    .appName("EcommerceML") \
    .getOrCreate()
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
file_path = "/content/Ecommerce_Customers.csv"

spark_df = spark.read.csv(
    file_path,
    header=True,
    inferSchema=True
)
```

```python
spark_df.show(5)
spark_df.printSchema()
```

```python
feature_columns = [
    'Avg Session Length',
    'Time on App',
    'Time on Website',
    'Length of Membership'
]
```

```python
assembler = VectorAssembler(
    inputCols=feature_columns,
    outputCol='features'
)

final_data = assembler.transform(spark_df)
```

**LINEAR** **REGRESSION**

```python
lr_data = final_data.select('features', 'Yearly Amount Spent')
```

```python
train_data, test_data = lr_data.randomSplit([0.8, 0.2], seed=42)
```

```python
lr = LinearRegression(
    featuresCol='features',
    labelCol='Yearly Amount Spent'
)

lr_model = lr.fit(train_data)
```

```python
lr_predictions = lr_model.transform(test_data)

lr_predictions.select(
    'prediction',
    'Yearly Amount Spent'
).show(10)
```

```python
reg_evaluator = RegressionEvaluator(
    labelCol='Yearly Amount Spent',
    predictionCol='prediction',
    metricName='r2'
)

r2_score = reg_evaluator.evaluate(lr_predictions)

print("R2 Score:", r2_score)
```

**LOGISTIC REGRESSION**

```python
classification_df = spark_df.withColumn(
    'label',
    when(col('Yearly Amount Spent') > 500, 1).otherwise(0)
)
```

```python
classification_data = assembler.transform(classification_df)
classification_data = classification_data.select('features', 'label')
```

```python
train_cls, test_cls = classification_data.randomSplit([0.8, 0.2], seed=42)
```

```python
log_reg = LogisticRegression(
    featuresCol='features',
    labelCol='label'
)

log_model = log_reg.fit(train_cls)
```

```python
log_predictions = log_model.transform(test_cls)

log_predictions.select(
    'prediction',
    'label'
).show(10)
```

```python
cls_evaluator = MulticlassClassificationEvaluator(
    labelCol='label',
    predictionCol='prediction',
    metricName='accuracy'
)

accuracy = cls_evaluator.evaluate(log_predictions)

print("Logistic Regression Accuracy:", accuracy)
```

**DECISION TREE CLASSIFIER**

```python
dt = DecisionTreeClassifier(
    featuresCol='features',
    labelCol='label'
)

dt_model = dt.fit(train_cls)
```

```python
dt_predictions = dt_model.transform(test_cls)

dt_predictions.select(
    'prediction',
    'label'
).show(10)
```

```python
dt_accuracy = cls_evaluator.evaluate(dt_predictions)

print("Decision Tree Accuracy:", dt_accuracy)
```

**RANDOM FOREST CLASSIFIER**

```python
rf = RandomForestClassifier(
    featuresCol='features',
    labelCol='label',
    numTrees=100
)

rf_model = rf.fit(train_cls)
```

```python
rf_predictions = rf_model.transform(test_cls)

rf_predictions.select(
    'prediction',
    'label'
).show(10)
```

```python
rf_accuracy = cls_evaluator.evaluate(rf_predictions)

print("Random Forest Accuracy:", rf_accuracy)
```

**K-MEANS CLUSTERING**



```python
cluster_data = assembler.transform(spark_df)
cluster_data = cluster_data.select('features')
```

```python
kmeans = KMeans(
    featuresCol='features',
    k=3,
    seed=42
)

kmeans_model = kmeans.fit(cluster_data)
```

```python
cluster_results = kmeans_model.transform(cluster_data)

cluster_results.show(10)
```

```python
centers = kmeans_model.clusterCenters()

for index, center in enumerate(centers):
    print(f"Cluster {index} Center: {center}")
```
